# Calculadora em Python

Projeto desenvolvido em setembro de 2025 durante o módulo de Lógica de programação com Python do curso de Data Analytics da EBAC.

## Objetivo

Praticar conceitos básicos de lógica de programação em Python, como entrada de dados, estruturas condicionais e loops.

## Operações

- Soma
- Subtração
- Multiplicação
- Divisão
- Potenciação

```python
operacao = input ("Qual operação você deseja realizar?\n1 - Soma\n2 - Subtração\n3 - Multiplicação\n4 - Divisão\n5 - Potência\n0 - Sair")

if operacao == "1":
    num1 = input ("Digite o primeiro número")
    num2 = input ("Digite o segundo número")
    num1 = float(num1)
    num2 = float(num2)
    soma = (num1 + num2)
    print (soma)

elif operacao == "2":
    num1 = input ("Digite o primeiro número")
    num2 = input ("Digite o segundo número")
    num1 = float(num1)
    num2 = float(num2)
    subtracao = (num1 - num2)
    print (subtracao)

elif operacao == "3":
    num1 = input ("Digite o primeiro número")
    num2 = input ("Digite o segundo número")
    num1 = float(num1)
    num2 = float(num2)
    multiplicacao = (num1 * num2)
    print (multiplicacao)

elif operacao == "4":
    num1 = input ("Digite o primeiro número")
    num2 = input ("Digite o segundo número")
    num1 = float(num1)
    num2 = float(num2)
    divisao = (num1 / num2)
    print (divisao)

elif operacao == "5":
    num1 = input ("Digite o primeiro número")
    num2 = input ("Digite o segundo número")
    num1 = float(num1)
    num2 = float(num2)
    potencia = (num1 ** num2)
    print (potencia)

elif operacao == "0":
    print ("Obrigado por usar a calculadora!")
    exit()

else:
    print ("Operação inválida!")
    exit()

while operacao != "0":
  sim_equivalencias = {"S", "SI", "SIM", "YES", "s"}
  nao_equivalencias = {"N", "NAO", "não", "no"}
  loop = input ("Você deseja continuar calculando? Digite SIM ou NÃO:")
  if loop.lower() in sim_equivalencias:
    operacao = input ("Qual operação você deseja realizar?\n1 - Soma\n2 - Subtração\n3 - Multiplicação\n4 - Divisão\n5 - Potência\n0 - Sair")
  elif loop.lower() in nao_equivalencias:
    print ("Obrigado por usar a calculadora!")
    exit()
  else:
    print ("Resposta inválida. Por favor, digite 'sim' ou 'nao'.")
```
